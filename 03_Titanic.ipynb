## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
pclass_survived = df.groupby('pclass')['survived'].mean() * 100
print("Tỷ lệ sống sót theo hạng vé (%):\n", pclass_survived)
diff = pclass_survived.max() - pclass_survived.min()
print(f"Chênh lệch giữa cao nhất và thấp nhất là: {diff:.2f}%")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
print("Tỷ lệ sống sót theo giới tính (%):\n", df.groupby('sex')['survived'].mean() * 100)
print("\nTỷ lệ sống sót theo hạng vé (%):\n", df.groupby('pclass')['survived'].mean() * 100)
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
fare_survived = df.groupby('survived')['fare'].mean()
print("Giá vé trung bình theo trạng thái sống sót:\n", fare_survived)
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
family_survived = df.groupby('family_size')['survived'].mean() * 100
print("Tỷ lệ sống sót theo quy mô gia đình (%):\n", family_survived)
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO
embark_survived = df.groupby('embark_town')['survived'].mean() * 100
print("Tỷ lệ sống sót theo cảng lên tàu (%):\n", embark_survived)
```
